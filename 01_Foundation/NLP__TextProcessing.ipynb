## Text Preprocessing

```python
sample_text="NLTK is a powerful library for natural language processing."

# Word tokenization
Word_token=nltk.word_tokenize(sample_text)
print("Word_token",Word_token)

# Sentence tokenize
Sentence_token=nltk.sent_tokenize(sample_text)
print("Sentence_token",Sentence_token)
```

```python
nltk.download('stopwords')

# Get English stop words
stop_words = set(stopwords.words('english'))

# Remove stop words from Word_token
filtered_words = [word for word in Word_token if word.lower() not in stop_words]

print("Original Words:", Word_token)
print("Filtered Words (after stop word removal):", filtered_words)
```

### Cleaning

```python
import re,string

text = "Hello!! Visit https://example.com for NLP tips... It's gr8!!! 😊"

#lower
text=text.lower()
print(text)


text=re.sub(r'[^\w\s]','',text)

# Remove URL
text=re.sub(r'https?://\S+','',text)
print("Removed URl--- ",text)

# Remove Punctuation
text=text.translate(str.maketrans('','',string.punctuation))
print("Removed Punctuation--- ",text)

# Remove Numbers
text=re.sub(r'\d+','',text)
print("Removed Numbers--- ",text)
# emove extra whitespace
text=''.join(text.split())
print("Removed extra whitespace--- ",text)

print(text)

```

### Named Entity Recognition (NER)

```python
from nltk import word_tokenize, pos_tag, ne_chunk

# Download the required resource for NER
nltk.download('maxent_ne_chunker_tab')
nltk.download('words') # This resource is also needed for the chunker

# Sample text
text = "Barack Obama was born in Hawaii in 1961."

# Tokenize and POS tag the sentence
tokens = word_tokenize(text)
tags = pos_tag(tokens)

# Apply Named Entity Recognition
entities = ne_chunk(tags)
print(entities)
```
